```python
import pandas as pd
import json

# read Excel
df = pd.read_excel("ACAI_ranked_universities_final.xlsx", sheet_name="ACAI_ranked_universities")

# columns need to process
cols_to_split = [
    "t1UniversityRulesPoliciesRegulations",
    "t2CenterForTeachingAndLearning",
    "t3AiInstituteInitiativeCenterHubAi",
    "t4LibraryLibraryGuides",
    "t5AcademicIntegrityHonorCode",
    "t6AiSteeringCommitteeTaskForce",
    "t7OtherRelevantLinks"
]

data_list = []

for _, row in df.iterrows():
    item = {}
    for col in df.columns:
        cell = row[col]
        if col in cols_to_split:
            if pd.notna(cell) and str(cell).strip():
                # split using \n，remove leading '- ' and trailing whitespace
                lines = [line.strip().lstrip("-").strip() for line in str(cell).split("\n") if line.strip()]
                item[col] = lines
            else:
                item[col] = []
        elif pd.notna(cell):
            item[col] = cell
        else:
            item[col] = None
    data_list.append(item)

# JSON output
with open("institutions_final.json", "w", encoding="utf-8") as f:
    json.dump(data_list, f, ensure_ascii=False, indent=2)
```
