```python
# ==================================================
# Imports
# ==================================================
import pandas as pd
import numpy as np
from scipy.stats import pearsonr, spearmanr, kendalltau

import matplotlib.pyplot as plt
import seaborn as sns
import matplotlib.colors as mcolors

# ==================================================
# Parameters
# ==================================================
SEED = 42
N_BOOT = 1000
rng = np.random.default_rng(SEED)

PLOT_TEMP = 1.0

LLM_COL = "ACAI_equal"
HUMAN_COL = "ACAI_equal"

# ==================================================
# Paths
# ==================================================
LLM_PATH = "../data/ACAI-US79-LLM-ACAI.csv"
HUMAN_PATH = "../data/ACAI-US79-scored.csv"
ATTR_PATH = "../data/ACAI_CSRankings_manually_matched.csv"

# ==================================================
# Helper functions
# ==================================================
def p_stars(p):
    if p < 0.001:
        return "***"
    elif p < 0.01:
        return "**"
    elif p < 0.05:
        return "*"
    return ""

def is_constant(arr):
    return np.unique(arr).size < 2

def text_color_for_bg(hex_color):
    r, g, b = mcolors.to_rgb(hex_color)
    luminance = 0.299*r + 0.587*g + 0.114*b
    return "black" if luminance > 0.6 else "white"

# ==================================================
# Subgroup statistics
# ==================================================
def subgroup_stats(df, grouping, xcol, ycol, n_boot, rng):
    rows = []

    for level, dfg in df.groupby(grouping):
        if len(dfg) < 5:
            continue

        x = dfg[xcol].values
        y = dfg[ycol].values

        if is_constant(x) or is_constant(y):
            continue

        pr, pp = pearsonr(x, y)
        sr, sp = spearmanr(x, y)
        kr, kp = kendalltau(x, y)

        boot = []
        for _ in range(n_boot):
            idx = rng.choice(len(x), len(x), replace=True)
            xb, yb = x[idx], y[idx]
            if is_constant(xb) or is_constant(yb):
                continue
            boot.append(spearmanr(xb, yb)[0])

        if len(boot) == 0:
            continue

        ci_low, ci_high = np.percentile(boot, [2.5, 97.5])

        rows.append({
            "group": level,
            "n": len(dfg),
            "pearson_r": pr,
            "pearson_p": pp,
            "pearson_sig": f"{pr:.2f}{p_stars(pp)}",
            "spearman_corr": sr,
            "spearman_p": sp,
            "spearman_sig": f"{sr:.2f}{p_stars(sp)}",
            "spearman_ci_low": ci_low,
            "spearman_ci_high": ci_high,
            "kendall_tau": kr,
            "kendall_p": kp,
            "kendall_sig": f"{kr:.2f}{p_stars(kp)}",
        })

    return pd.DataFrame(rows)

# ==================================================
# Load data
# ==================================================
df_llm = pd.read_csv(LLM_PATH)
df_human = pd.read_csv(HUMAN_PATH)
df_attr = pd.read_csv(ATTR_PATH)

df_attr = df_attr[
    ["Index", "Type", "Research Activity", "Regional Coverage", "size_bucket"]
]

# ==================================================
# Loop over temperatures
# ==================================================
SUBGROUPS = [
    "Type",
    "Regional Coverage",
    "Research Activity",
    "size_bucket",
]

subgroup_results_by_temp = {}

for temp in sorted(df_llm["run_temperature"].unique()):
    print("\n" + "="*60)
    print(f" OVERALL LLM vs HUMAN AGREEMENT — Temperature = {temp}")
    print("="*60)

    df = (
        df_llm[df_llm["run_temperature"] == temp][["Index", LLM_COL]]
        .merge(df_human[["Index", HUMAN_COL]], on="Index",
               suffixes=("_llm", "_human"))
        .merge(df_attr, on="Index", how="left")
        .dropna()
    )

    x = df[f"{LLM_COL}_llm"]
    y = df[f"{HUMAN_COL}_human"]

    pr, pp = pearsonr(x, y)
    sr, sp = spearmanr(x, y)
    kr, kp = kendalltau(x, y)

    # Bootstrap CI
    boot = []
    for _ in range(N_BOOT):
        idx = rng.choice(len(x), len(x), replace=True)
        xb, yb = x.values[idx], y.values[idx]
        if is_constant(xb) or is_constant(yb):
            continue
        boot.append(spearmanr(xb, yb)[0])

    ci_low, ci_high = np.percentile(boot, [2.5, 97.5])

    # Rank gaps
    df["rank_llm"] = x.rank(ascending=False, method="first") # when there's a tie it's the first occurring
    df["rank_human"] = y.rank(ascending=False, method="first")
    df["rank_gap"] = (df["rank_llm"] - df["rank_human"]).abs()

    print(f"n universities            : {len(df)}")
    print(f"Pearson r                 : {pr:.2f}{p_stars(pp)}")
    print(f"Spearman r                : {sr:.2f}{p_stars(sp)}")
    print(f"Kendall tau               : {kr:.2f}{p_stars(kp)}")
    print(f"Spearman 95% CI           : [{ci_low:.2f}, {ci_high:.2f}]")
    print(f"Mean |Δ rank|             : {df['rank_gap'].mean():.2f}")
    print(f"Median |Δ rank|           : {df['rank_gap'].median():.2f}")
    print(f"Min |Δ rank|              : {df['rank_gap'].min():.2f}")
    print(f"Max |Δ rank|              : {df['rank_gap'].max():.2f}")

    # ==================================================
    # Bottom / Top 25% rank-gap analysis (WITH MIN/MAX)
    # ==================================================
    q25, q75 = df["rank_gap"].quantile([0.25, 0.75])

    low_gap = df[df["rank_gap"] <= q25]
    high_gap = df[df["rank_gap"] >= q75]

    print("\n--- Rank-gap tails (descriptive) ---")

    print(
        f"Bottom 25% (closest agreement): "
        f"n={len(low_gap)}, "
        f"|Δ rank| ∈ [{low_gap['rank_gap'].min():.1f}, {low_gap['rank_gap'].max():.1f}]"
    )

    print(
        f"Top 25% (largest disagreement): "
        f"n={len(high_gap)}, "
        f"|Δ rank| ∈ [{high_gap['rank_gap'].min():.1f}, {high_gap['rank_gap'].max():.1f}]"
    )

    def summarize_tail(dft, label):
        print(f"\n[{label}]")
        display(
            dft[["Type", "Research Activity", "Regional Coverage", "size_bucket"]]
            .value_counts()
            .rename("count")
            .reset_index()
        )

    summarize_tail(low_gap, "Bottom 25% |Δ rank|")
    summarize_tail(high_gap, "Top 25% |Δ rank|")

    # ---- Subgroups (correlations) ----
    print("\n--- Exploratory subgroup correlations ---")

    temp_tables = {}
    for g in SUBGROUPS:
        tbl = subgroup_stats(
            df,
            grouping=g,
            xcol=f"{LLM_COL}_llm",
            ycol=f"{HUMAN_COL}_human",
            n_boot=N_BOOT,
            rng=rng
        ).round(3)

        temp_tables[g] = tbl
        print(f"\n[{g}]")
        display(tbl)

    subgroup_results_by_temp[temp] = temp_tables

# ==================================================
# Color, label, and order maps
# ==================================================
COLOR_MAP = {
    "Type": {
        "Public Research-Oriented": "#B7D4F0",
        "Private Research-Oriented": "#A9DED6",
        "Teaching-Oriented/Liberal Arts": "#E0CEF2"
    },
    "Research Activity": {
        "R1": "#A9DDB7",
        "R2": "#E6F0A3",
        "-": "#E2E2E2"
    },
    "Regional Coverage": {
        "South": "#F0C1B1",
        "Midwest": "#E6D39B",
        "Northeast": "#C7D9F0",
        "West": "#C6E2CE"
    },
    "size_bucket": {
        "small": "#F2C7BC",
        "medium": "#C8D4F0",
        "large": "#C6B9EA"
    }
}

LABEL_MAP = {
    "size_bucket": {"small": "Small", "medium": "Medium", "large": "Large"},
    "Research Activity": {"R1": "R1", "R2": "R2", "-": "None"},
    "Regional Coverage": {
        "Northeast": "Northeast", "Midwest": "Midwest",
        "South": "South", "West": "West"
    },
    "Type": {
        "Teaching-Oriented/Liberal Arts": "Teaching-Oriented/Liberal Arts",
        "Private Research-Oriented": "Private Research-Oriented",
        "Public Research-Oriented": "Public Research-Oriented",
    },
}

ORDER_MAP = {
    "size_bucket": {"small": 0, "medium": 1, "large": 2},
    "Research Activity": {"R1": 0, "R2": 1, "-": 2},
    "Type": {
        "Teaching-Oriented/Liberal Arts": 0,
        "Private Research-Oriented": 1,
        "Public Research-Oriented": 2,
    },
    "Regional Coverage": {
        "Northeast": 0, "Midwest": 1, "South": 2, "West": 3,
    },
}

# ==================================================
# Forest plot (TEMP = 1.0)
# ==================================================
sns.set_theme(style="whitegrid", context="talk")

plot_df = results_df.copy()
plot_df["order"] = plot_df.apply(
    lambda r: ORDER_MAP.get(r["grouping"], {}).get(r["group"], 999), axis=1
)

plot_df = plot_df.sort_values(["grouping", "order"]).reset_index(drop=True)

plot_df["pretty_label"] = plot_df.apply(
    lambda r: f"{LABEL_MAP.get(r['grouping'], {}).get(r['group'], r['group'])} (n={r['n']})",
    axis=1
)

fig, ax = plt.subplots(figsize=(23, 6.5))

ax.hlines(
    plot_df.index,
    plot_df["spearman_ci_low"],
    plot_df["spearman_ci_high"],
    linewidth=4,
    alpha=0.4
)

sns.scatterplot(
    data=plot_df,
    x="spearman_corr",
    y=plot_df.index,
    size="n",
    sizes=(80, 500),
    legend=False,
    ax=ax,
    zorder=3
)

ax.axvline(0, linestyle="--", alpha=0.5)
ax.axvspan(-0.30, 0.30, color="grey", alpha=0.15)

ax.set_yticks(plot_df.index)
ax.set_yticklabels(plot_df["pretty_label"])

for tick, (_, row) in zip(ax.get_yticklabels(), plot_df.iterrows()):
    bg = COLOR_MAP.get(row["grouping"], {}).get(row["group"], "#FFFFFF")
    tick.set_bbox(dict(facecolor=bg, boxstyle="round,pad=0.4", alpha=0.9))
    tick.set_color(text_color_for_bg(bg))

ax.set_xlabel("Spearman Correlation with 95% CI (Temperature = 1.0)")
ax.set_ylabel("")

sns.despine(left=True, bottom=True)
plt.tight_layout(rect=(0.32, 0.05, 1, 1))
plt.show()
```
